- `pip install matplotlib` komutu ile matplotlib kütüphanesini kurabiliriz.

### `plt.plot()`
ile grafiği çizdirir
### `plt.show()`
ile grafiği gösterir

```python
import numpy as np
import matplotlib.pyplot as plt
x = np.array([1,8])
y = np.array([0,150])
plt.plot(x,y)
plt.show()
```

- plot içinde "o" yazarsak noktaları gösterir. 

```python
plt.plot(x,y, "o")
plt.show()
```

- plot içersinde marker parametresi ile kırılma noktalarını gösteririz.

```python
x = np.array([2,4,6,8,10])
y = np.array([1,3,5,7,9])
plt.plot(x,y)
plt.show()
```

```python
plt.plot(x,y, "o")
plt.show()
```

- marker parametresi ile nokta yerine istediğimiz şekli gösterebiliriz.

```python
plt.plot(x,y, marker = "o")
```

```python
plt.plot(x,y, marker = "*")
```

- linestyle parametresi ile çizgi türünü belirleriz.

```python
y = np.array([13,20,11,100])
plt.plot(y, linestyle="dashdot", color="r")
plt.show()
```

- çoklu grafik çizdirme

```python
plt.plot(y)
plt.title("Bu ana başlık")
plt.xlabel("X ekseni")
plt.ylabel("Y ekseni")
plt.grid()
plt.show()
```

## Subplots (Çoklu grafik)

```python
x = np.array([80,85,90,100,105,110])
y = np.array([240,250,260,270,280,290])
plt.subplot(1,2,1)
plt.title("1")
plt.plot(x,y)
plt.show()

x = np.array([80,85,90,100,105,110])
y = np.array([240,250,260,270,280,290])
plt.subplot(1,2,2)
plt.title("2")
plt.plot(x,y)
plt.show()
```
